## Q4. Create a DataFrame using a NumPy array of your choice.

```python
# Write your answer here
import numpy as np

data = np.array([

            ['Chetan', 'Rohan', 'Ramesh', 'Niraj', 'Piyush'],
            [19, 21, 23, 20, 10],
            ['Nagpur', 'Bhopal', 'Indore', 'Sanchi', 'Delhi']

        ])
df = pd.DataFrame(data, index=['Name', 'Age', 'City'])
print(df, '\n')


name = np.array(['Chetan', 'Rohan', 'Ramesh', 'Niraj', 'Piyush'])
age = np.array([19, 21, 23, 20, 10])
city = np.array(['Nagpur', 'Bhopal', 'Indore', 'Sanchi', 'Delhi'])

df = pd.DataFrame({
    'Name' : name,
    'Age' : age,
    'City' : city
})
print(df)
```
