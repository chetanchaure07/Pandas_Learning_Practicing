### Q36. Create a pivot table showing **mean Salary by Department and City** using `pivot_table()`.

# Q40. ⭐ Final Practical Challenge

Using the same 8-column DataFrame:

1. Sort employees by `Salary` from highest to lowest.
2. Find the average Salary of each Department.
3. Find the Department with the highest average Salary.
4. Find the average Score by City.
5. Create a pivot table of mean Salary by Department and City.
6. Create a crosstab of Department vs Gender.

**Expected Result:** A small analysis report containing all six results.

```python
# Write your final solution here
print(df.sort_values('Salary', ascending = False), '\n')

print(df.groupby('Department')['Salary'].mean(), '\n')

print(df.groupby('Department')['Salary'].mean().sort_values(ascending = False), '\n')

print(df.groupby('City')['Score'].mean(), '\n')

pivot = pd.pivot_table(
    df,
    values = 'Salary',
    index = 'Department',
    columns = 'City',
    aggfunc = 'mean'
)
print(pivot, '\n')

print(pd.crosstab(df['Department'], df['Gender']))
```
