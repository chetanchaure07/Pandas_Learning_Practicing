## 1. Using `concat()` for Stacking Data
- `pd.concat()` combines DataFrames vertically (rows) or horizontally (columns).
- Use `axis=0` for vertical stacking (default), `axis=1` for horizontal.
- `ignore_index=True` can reset the index after concatenation.

```python

# Dataset 1: Product details

df1 = pd.DataFrame({
    "ProductID" : [101, 102, 103],
    "ProductName" : ["Pen", "Notebook", "Marker"],
    "Price" : [10, 20, 30]
})
print(df1, "\n")
# Dataset 2: More products (same columns as df1 → good for vertical stacking)
df2 = pd.DataFrame({
    "ProductID" : [104, 105],
    "ProductName" : ["Pencil", "Erase"],
    "Price" : [5, 8]
})
print(df2, "\n")

# Dataset 3: Additional info (different columns → good for horizontal stacking)
df3 = pd.DataFrame({
    "Stock" : [100, 60, 80],
    "Rating" : [4.5, 4.7, 4.8]
})


# 1. Vertical Stacking (Works on rows) -> combine df1 and df2
print(pd.concat([df1, df2], ignore_index = True), '\n')

# 2. Horizontal Stacking (Works on columns df1 and df2)
print(pd.concat([df1, df3], axis = 1))

# axis is by default = 0 (rows)
# axis is set = 1 (columns)
```

## 2. Using `merge()` for SQL-Style Joins
- `merge()` is similar to SQL joins.
- Specify `on` or `left_on`/`right_on` to choose key columns.
- Join types: `inner` (default), `left`, `right`, `outer`.

```python
# Products = concatenting df1 and df2 on this basis of rows
products = pd.concat([df1, df2], ignore_index=True)

sales = pd.DataFrame({
    "ProductID" : [101, 102, 104, 106],
    "UnitsSold" : [500, 200, 150, 50],
})

print(products, '\n')
print(sales, '\n')

# Inner Join (Only matching ProductID)
inner = pd.merge(products, sales, on="ProductID", how="inner")
print(inner, '\n')

# Outer Join (Join Just like union)
outer = pd.merge(products, sales, on="ProductID", how="outer")
print(outer, '\n')

# Left Join (All elements from left and common from righ)
left = pd.merge(products, sales, on="ProductID", how="left")
print(left, '\n')

# Right Join (All elements from right and common from left)
right = pd.merge(products, sales, on="ProductID", how="right")
print(right)

# Merge focuses on column and join focuses on index
```

## 3. Using `join()` with Index Alignment
- `join()` joins DataFrames on their index or on a key column.
- By default, joins on index; can specify `on='col'`.
- Useful for adding columns to an existing DataFrame.

```python
print(products, '\n')
stock = pd.DataFrame({
    "Stock" : [100, 60, 80, 120, 150],
},  index = [101, 102, 103, 104, 105])
print(stock, '\n')

print("Merge focuses on column and join focuses on index\n")

# products = products.set_index("ProductID")
print(products, '\n')
print(products.join(stock), '\n')

# We can still join two df instead of using set_index
print(products.join(stock, on='ProductID'))
```

## 4. Practical Examples
- Combine sales and revenue data to see performance.
- Merge students with marks to link records.
- Practice with realistic datasets to understand alignment and missing data handling.

```python
import pandas as pd
```

```python
sales = pd.DataFrame({
    "product": ["A", "B", "C"],
    "sales": [100, 150, 200]
})

revenue = pd.DataFrame({
    "product": ["A", "B", "D"],
    "revenue": [10000, 18000, 12000]
})

result = pd.merge(sales, revenue, on='product', how='outer')
print(result)
```

```python
students = pd.DataFrame({
    "student_id": [1, 2, 3, 4],
    "name": ["Amit", "Priya", "Rahul", "Sneha"]
})

marks = pd.DataFrame({
    "student_id": [1, 2, 4],
    "marks": [85, 92, 78]
})

result = pd.merge(students, marks, on = 'student_id', how='left')
result
```

```python
result.isna().sum()
result['marks'] = result['marks'].fillna(0)
result
```
