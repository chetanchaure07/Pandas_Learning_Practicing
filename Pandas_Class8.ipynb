## 1. Converting Columns to Datetime (`pd.to_datetime`)
- Use `pd.to_datetime()` to convert strings or numbers into datetime objects.
- Once converted, Pandas recognizes them as dates, enabling powerful operations.

```python
import pandas as pd

# 1. Create the dataset
data = {
    "OrderID": [1001, 1002, 1003, 1004, 1005],
    "OrderDate": ["2024-01-05", "2024/01/12", "15-01-2024", "2024-01-20", "20240125"],
    "Customer": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Amount": [250, 400, 150, 300, 500]
}

df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
```

```python
a = pd.Series(['10', '20', '30', 'Chacha', 'Modiji', '50'], name = 'Sample')
print(a, '\n')

# If we want to convert the actual numerical value in data type of intand forget about those
# Which are not in numerical format.
print(pd.to_numeric(a, errors='coerce'))
```

```python
print(pd.to_datetime(df['OrderDate'], errors="coerce"))
```

```python
# We are fixing all the format of the date which is not in the correct
# Format of %Y-%m-%d

df['OrderDate'] = pd.to_datetime(df['OrderDate'], format="mixed")
print(df)

# df['OrderDate'].dt.strftime("%d-%Y-%m")
```

## 2. Setting DateTime Index & Resampling
- Set a datetime column as the index with `set_index()`.
- Resample time series using `resample('M').mean()` or other rules ('D', 'W', 'Y').
- Useful for aggregating data by time periods.

```python
df = pd.DataFrame({
    'date': ['2025-01-01', '2025-01-15', '2025-02-01', '2025-02-20'],
    'sales': [100, 150, 200, 250]
})

# Convert the column to datetime
df['date'] = pd.to_datetime(df['date'])

# Set datetime column as index
df = df.set_index('date')

# Monthly average
monthly_avg = df.resample('ME').mean()

print(monthly_avg)
```

## mini-project

Converting columns to datetime (pd.to_datetime)

Setting a DateTime index & resampling

```python
import pandas as pd

# 1️⃣ Create a dataset of website visits
data = {
    "VisitDate": [
        "2026-01-05", "2026-01-12", "2026-02-03",
        "2026-02-18", "2026-03-10", "2026-03-25",
        "2026-04-07", "2026-04-20", "2026-05-15"
    ],
    "Visits": [100, 150, 200, 180, 250, 300, 220, 270, 350]
}

df = pd.DataFrame(data)

# 2️⃣ Convert VisitDate to datetime
df["VisitDate"] = pd.to_datetime(df["VisitDate"])

# 3️⃣ Set VisitDate as the index
df.set_index("VisitDate", inplace=True)

# 4️⃣ Resample to get total visitors per month
monthly_visits = df["Visits"].resample("ME").sum()
print("Total visitors per month:")
print(monthly_visits)

# 5️⃣ Resample to get average visitors per week
weekly_visits = df["Visits"].resample("W").mean()
print("\nAverage visitors per week:")
print(weekly_visits)
```
