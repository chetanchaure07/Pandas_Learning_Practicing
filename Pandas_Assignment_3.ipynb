### Q13. Fill missing values of Age with 20.

```python
# Write your code here
df['Age'] = df['Age'].fillna(20)
df['Age']
```

### Q14. Fill missing values of Marks using the column mean.

```python
# Write your code here
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
print(df['Marks'].mean())
df['Marks']
```

### Q15. Remove duplicate rows.

```python
# Write your code here
print('Total Duplicate Values : ', df.duplicated().sum(), '\n')
df = df.drop_duplicates(subset='Name', keep="first", ignore_index=True)
df
```

### Q16. Rename the column Marks to Score.

```python
# Write your code here
df = df.rename(columns={'Marks' : 'Score'})
df
```
