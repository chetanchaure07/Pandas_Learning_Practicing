## Q1. Create a DataFrame named `sales`:

| Date | Product | Sales |
|---|---|---:|
| 2024-02-01 | Laptop | 50000 |
| 2024/02/03 | Mouse | 1200 |
| 05-02-2024 | Keyboard | 2500 |
| 2024-02-10 | Monitor | 15000 |
| 20240215 | Laptop | 52000 |
| 2024-02-18 | Mouse | 1400 |
| 20-02-2024 | Keyboard | 2800 |
| 2024-02-25 | Monitor | 16000 |
| 2024/02/27 | Laptop | 48000 |
| 2024-02-29 | Mouse | 1300 |

Expected Output: 10 rows × 3 columns.

```python
# Write your answer here
sales = pd.DataFrame({
    'date' : ["2024-02-01", "2024/02/03", "05-02-2024", "2024-02-10", "20240215", "2024-02-18", "20-02-2024", "2024-02-25", "2024/02/27", "2024-02-29"],
    'product' : ["Laptop", "Mouse", "Keyboard", "Monitor", "Laptop", "Mouse", "Keyboard", "Monitor", "Laptop", "Mouse"],
    'sales' : [50000, 1200, 2500, 15000, 52000, 1400, 2800, 16000, 48000, 1300]
})
sales
```

## Q2. Convert `Date` to datetime using `pd.to_datetime()`.

Expected Output: Date has datetime dtype.

```python
# Write your answer here
sales['date'] = pd.to_datetime(sales['date'], format='mixed')

print(sales.dtypes)
```

## Q3. Find the earliest and latest dates.

Expected Output: Minimum and maximum dates are displayed.

```python
# Write your answer here
earliest_date = sales['date'].min()
latest_date = sales['date'].max()

print("Earliest date:", earliest_date)
print("Latest date:", latest_date)
```

## Q4. Filter records after `2024-02-10`.

Expected Output: Only records after February 10.

```python
# Write your answer here
filtered_sales = sales[sales['date'] > '2024-02-10']

print(filtered_sales)
```

## Q5. Filter records between `2024-02-05` and `2024-02-20`.

Expected Output: Records in that date range.

```python
# Write your answer here
filtered_sales = sales[
    sales['date'].between('2024-02-05', '2024-02-20')
]

print(filtered_sales)
```

## Q6. Set `Date` as the DataFrame index.

Expected Output: Date becomes a DateTimeIndex.

```python
# Write your answer here
sales['date'] = pd.to_datetime(sales['date'], format='mixed')

sales = sales.set_index('date')

print(sales)
print(type(sales.index))
```

## Q7. Using the DateTime index, select all February 2024 records.

Expected Output: All February records.

```python
# Write your answer here
feb_sales = sales.loc['2024-02']

print(feb_sales)
```

## Q8. Select records from `2024-02-10` to `2024-02-20` using date slicing.

Expected Output: Records in that range.

```python
# Write your answer here
sales = sales.sort_index()

sales.loc['2024-02-10':'2024-02-20']
```

## Q9. Resample by week and calculate total Sales.

Expected Output: Weekly total Sales.

```python
# Write your answer here
weekly_sales = sales['sales'].resample('W').sum()

print(weekly_sales)
```

## Q10. Resample by month and calculate total Sales.

Expected Output: One monthly total for February.

```python
# Write your answer here
monthly_sales = sales['sales'].resample('ME').sum()

print(monthly_sales)
```

## Q11. Resample by month and calculate average Sales.

Expected Output: One monthly average.

```python
# Write your answer here
monthly_avg_sales = sales['sales'].resample('ME').mean()

print(monthly_avg_sales)
```

## Q12. Resample by day and calculate total Sales.

Expected Output: Daily totals.

```python
# Write your answer here
daily_sales = sales['sales'].resample('D').sum()

print(daily_sales)
```

## Q13. Calculate a 3-observation rolling average of Sales using `rolling()`.

Expected Output: Rolling average series; initial values may be NaN.

```python
# Write your answer here
rolling_avg = sales['sales'].rolling(3).mean()

print(rolling_avg)
```

## Q14. Calculate a 3-observation rolling sum of Sales.

Expected Output: Rolling 3-observation totals.

```python
# Write your answer here
rolling_sum = sales['sales'].rolling(3).sum()

print(rolling_sum)
```

## Q15. Calculate the expanding average using `expanding().mean()`.

Expected Output: Cumulative average after each observation.

```python
# Write your answer here
expanding_avg = sales['sales'].expanding().mean()

print(expanding_avg)
```

## Q16. Calculate the expanding sum using `expanding().sum()`.

Expected Output: Cumulative Sales total.

```python
# Write your answer here
expanding_sum = sales['sales'].expanding().sum()

print(expanding_sum)
```

## Q17. Find the week with the highest total Sales after weekly resampling.

Expected Output: One week and its total Sales.

```python
# Write your answer here
weekly_sales = sales['sales'].resample('W').sum()

highest_week = weekly_sales.idxmax()
highest_sales = weekly_sales.max()

print("Week:", highest_week)
print("Total Sales:", highest_sales)
```

## Q18. Create a monthly summary with both total Sales and average Sales.

Expected Output: One row for February with two statistics.

```python
# Write your answer here
monthly_summary = sales['sales'].resample('ME').agg(['sum', 'mean'])

print(monthly_summary)
```

## Q19. Find the date on which the highest individual Sales value occurred.

Expected Output: One date and its Sales value.

```python
# Write your answer here
highest_date = sales['sales'].idxmax()
highest_sales = sales['sales'].max()

print("Date:", highest_date)
print("Sales:", highest_sales)
```

# ⭐ Bonus Challenge

Create your own 15-day sales dataset with `Date`, `Product`, and `Sales`.

Then:
- Convert Date to datetime.
- Set Date as index.
- Find weekly Sales.
- Calculate a 3-day rolling average.
- Calculate an expanding maximum.
- Find the best-performing week.

**Goal:** Perform the complete time-series workflow on your own dataset.

```python
# Bonus solution
import pandas as pd

# Create dataset
sales = pd.DataFrame({
    'Date': [
        '2024-03-01', '2024-03-02', '2024-03-03',
        '2024-03-04', '2024-03-05', '2024-03-06',
        '2024-03-07', '2024-03-08', '2024-03-09',
        '2024-03-10', '2024-03-11', '2024-03-12',
        '2024-03-13', '2024-03-14', '2024-03-15'
    ],
    'Product': [
        'Laptop', 'Mouse', 'Keyboard',
        'Monitor', 'Laptop', 'Mouse',
        'Keyboard', 'Monitor', 'Laptop',
        'Mouse', 'Keyboard', 'Monitor',
        'Laptop', 'Mouse', 'Monitor'
    ],
    'Sales': [
        45000, 1200, 2500,
        14000, 50000, 1500,
        2800, 16000, 48000,
        1300, 3000, 15500,
        52000, 1400, 17000
    ]
})

# Convert Date to datetime
sales['Date'] = pd.to_datetime(sales['Date'])

# Set Date as index
sales = sales.set_index('Date').sort_index()

# Weekly Sales
weekly_sales = sales['Sales'].resample('W').sum()

# 3-day rolling average
rolling_avg = sales['Sales'].rolling(3).mean()

# Expanding maximum
expanding_max = sales['Sales'].expanding().max()

# Best-performing week
best_week = weekly_sales.idxmax()
best_week_sales = weekly_sales.max()

print("Weekly Sales:")
print(weekly_sales)

print("\n3-Day Rolling Average:")
print(rolling_avg)

print("\nExpanding Maximum:")
print(expanding_max)

print("\nBest-performing Week:")
print("Week:", best_week)
print("Sales:", best_week_sales)
```
