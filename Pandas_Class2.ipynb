## 1 Inspecting Data
Quick ways to understand your dataset:
- `head(n)` → first *n* rows (default 5)
- `tail(n)` → last *n* rows
- `info()` → columns, non-null counts, and dtypes
- `shape` → (rows, columns)
- `dtypes` → data type of each column

```python
# Basic Inspection.
```

```python
import pandas as pd

# Create a small DataFrame
a = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Age": [25, 30, 35, 40, 22],
    "Score": [88, 92, 79, 85, 95]
}

b = pd.DataFrame(a)
print(b, "\n")

# 1. If we want to see only starting 5 rows
print(b.head(2), "\n")

# If b.head() is empty, it returns the default value specified in that line of code.

# 2. If we want to look up from the bottom or from the end
print(b.tail(2), "\n")

# 3. If we want to see about the info of the dataframe
print(b.info(), "\n")

# 4. Shape of the data
print(b.shape, "\n")

# 5. To check type of column
print(b.dtypes, "\n")
```

## 2 Selecting Columns & Rows
Ways to access parts of a DataFrame:
- Single column → `df['col']`
- Multiple columns → `df[['col1','col2']]`
- Label-based selection → `df.loc[row_label, col_label]`
- Integer-index selection → `df.iloc[row_idx, col_idx]`
Tip: `:` means "all rows/columns" in that dimension.

```python
print(b, "\n")

# 1. Selecting only one column (Age)
print(b['Age'], "\n")

# 2. Selecting multiple columns
# ---> We have to access using list of list
print(b[['Age', 'Score']], '\n')

# 3. Label Based Selection
print("Age of Charlie", b.loc[2, 'Age'])
print("Score of Emma", b.loc[4, 'Score'])
print("Age and Score of Alice", b.loc[1, ['Age', 'Score']], "\n")
# ---> Slicing in locaction
print(b.loc[2:3, ['Name', 'Score']], "\n")

# 4. iloc() Index Location
print(b.iloc[3], "\n") # row starting from index 1 to 2
print(b.iloc[1:3, :], "\n")  # Fir Filtering...
print(b.iloc[1:3, 1:3])
```

## 3 Index & Columns Overview
- `df.index` → row labels
- `df.columns` → column names
- `df.values` → underlying NumPy array (read-only; prefer using DataFrame ops)

```python
print(b, "\n")

# It shows start, stop and step value
print(b.index, "\n")

# Name of the columns
print(b.columns, "\n")

# DataFrame --> Matrix form
print(b.values, "\n")
```

## 4 Renaming Columns & Setting an Index
- Rename columns → `df.rename(columns={'old':'new'}, inplace=False)`
- Set a column as index → `df.set_index('column_name', inplace=False)`
- Reset index → `df.reset_index(inplace=False)`
 If you don't use `inplace=True`, remember to assign the result back to a variable.

```python
print(b, "\n")

print(b.rename(columns = {'Score' : 'Marks'}), "\n")

b['Index'] = ['A', 'B', 'C', 'D', 'E']
print(b, "\n")

print(b.set_index('Index', inplace = True), "\n")
```

## 5 Basic Attributes Recap
- `df.columns` → names of all columns
- `df.index` → row labels
- `df.values` → NumPy array (data only)
These are handy for quick checks and loops (though vectorized ops are preferred).
