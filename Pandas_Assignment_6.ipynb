### Q13. Create `employees` with `Emp_ID, Name` and `salary` with `Employee_ID, Salary`. Use IDs 1,2,3 in employees and 1,2,4 in salary. Merge using `left_on` and `right_on`.

**Expected Output: DataFrames merge using differently named key columns.**

```python
# Write your answer here
employees = pd.DataFrame({
    "Emp_ID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"]
})

salary = pd.DataFrame({
    "Employee_ID": [1, 2, 4],
    "Salary": [50000, 60000, 70000]
})

result = pd.merge(
    employees,
    salary,
    left_on="Emp_ID",
    right_on="Employee_ID",
    how="left"
)

print(result)
```

### Q14. Using the same employees and salary DataFrames, perform a left merge.

**Expected Output: All employees remain.**

```python
# Write your answer here
result = pd.merge(
    employees,
    salary,
    left_on="Emp_ID",
    right_on="Employee_ID",
    how="left"
)

print(result)
```

# ⭐ Bonus Challenge

Create three small DataFrames representing **Employees, Departments, and Salaries**.

- Combine two using `merge()`.
- Add the third using `join()`.
- Use an outer merge.
- Find missing records.
- Finally, explain why the missing values appeared.

**Goal:** Use `concat()`, `merge()`, and `join()` in one mini analysis.

```python
# Bonus solution
import pandas as pd

# 1. Create Employees DataFrame
employees = pd.DataFrame({
    "Emp_ID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"]
})

# 2. Create Departments DataFrame
departments = pd.DataFrame({
    "Emp_ID": [1, 2, 4],
    "Department": ["HR", "IT", "Finance"]
})

# 3. Create Salaries DataFrame
salaries = pd.DataFrame({
    "Emp_ID": [1, 2, 3, 5],
    "Salary": [50000, 60000, 55000, 70000]
})


# --- CONCAT ---
# Combine Employees and Departments vertically
concat_result = pd.concat([employees, departments], ignore_index=True)

print("CONCAT RESULT:")
print(concat_result)


# --- OUTER MERGE ---
# Combine Employees and Departments using Emp_ID
merged = pd.merge(
    employees,
    departments,
    on="Emp_ID",
    how="outer"
)

print("\nOUTER MERGE RESULT:")
print(merged)


# --- JOIN ---
# Set Emp_ID as index before joining Salaries
merged_indexed = merged.set_index("Emp_ID")
salaries_indexed = salaries.set_index("Emp_ID")

final_result = merged_indexed.join(salaries_indexed, how="outer")

print("\nFINAL RESULT:")
print(final_result)


# --- FIND MISSING RECORDS ---
print("\nMISSING VALUES:")
print(final_result.isna())

print("\nROWS WITH MISSING VALUES:")
print(final_result[final_result.isna().any(axis=1)])
```
